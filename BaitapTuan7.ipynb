### W7A1
#### Viết hàm tìm kiếm nhị phân trên một danh sách đã được sắp xếp tăng dần. Trả về chỉ số của phần tử hoặc -1 nếu không tìm thấy. mẫu hàm: def binary_search(arr, left, right, target):

```python
def binary_search(arr, left, right, target):
    while left<=right:
        mid=left+(right-left)//2
        if arr[mid]==target:
            return mid
        elif arr[mid]<target:
            left=mid+1
        else:
            right=mid-1
    return -1
data=list(map(int,input().split()))
target=data.pop()
print(binary_search(data, 0, len(data)-1, target))
```

### W7A3
#### Viết hàm thực hiện thuật toán bubble sort, Selection Sort để sắp xếp danh sách theo thứ tự tăng dần. Sắp xếp nổi bọt (Bubble Sort);Sắp xếp chọn (Selection Sort)

```python
def bubble_sort(arr):
    n=len(arr)
    for i in range(n):
        for j in range(0,n-i-1):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
    return arr
arr=list(map(int,input().split()))
z=bubble_sort(arr)
print(z)
```

```python
def selection_sort(arr):
    n=len(arr)
    for i in range(n):
        min=i
        for j in range(i+1,n):
            if arr[j]<arr[min]:
                min=j
        arr[i],arr[min]=arr[min],arr[i]
    return arr
arr=list(map(int,input().split()))
z=selection_sort(arr)
print(z)
```

### W7A7
#### Cho một danh sách các số nguyên nums, hãy tìm độ dài của dãy con tăng nghiêm ngặt dài nhất (không nhất thiết phải liền kề).


```python
def h(lst):
    if not lst:
        return 0
    a=[1]*len(lst)
    for i in range(1,len(lst)):
        for j in range(i):
            if lst[i]>lst[j]:
                a[i]=max(a[i],a[j]+1)
    return max(a)
lst=list(map(int,input().split()))
print(h(lst))
```

### W7A8
#### Bạn được cung cấp: một danh sách các khoảng thời gian interval = [[l1, r1], [l2, r2], ...] và một danh sách các truy vấn queries = [q1, q2, ...]
#### Với mỗi truy vấn q, hãy tìm độ dài nhỏ nhất của khoảng thời gian [li, ri] sao cho li ≤ q ≤ ri.
#### Nếu không tồn tại khoảng thời gian nào như vậy, hãy trả về -1 cho truy vấn đó. solve in < O(m.n), hint: binary search in the queries list


```python
import bisect

def minInterval(intervals, queries):
    # 1. Gắn vị trí gốc (index) vào từng query để trả kết quả đúng thứ tự
    # Ví dụ: queries = [2, 1, 3] -> [(1, 1), (2, 0), (3, 2)]
    sorted_queries = sorted([(q, i) for i, q in enumerate(queries)])
    
    # Tách riêng mảng giá trị để dùng hàm bisect tìm kiếm nhị phân
    q_vals = [q for q, i in sorted_queries]
    
    # 2. Sắp xếp các intervals theo ĐỘ DÀI TĂNG DẦN
    # Độ dài được tính bằng công thức: r - l + 1
    intervals.sort(key=lambda x: x[1] - x[0] + 1)
    
    # Tạo mảng kết quả mặc định là -1
    ans = [-1] * len(queries)
    
    # 3. Duyệt qua từng interval (từ ngắn nhất đến dài nhất)
    for l, r in intervals:
        length = r - l + 1
        
        # Dùng tìm kiếm nhị phân để tìm mốc bắt đầu và kết thúc của queries lọt vào [l, r]
        # - bisect_left tìm index đầu tiên có giá trị >= l
        # - bisect_right tìm index đầu tiên có giá trị > r, nên ta trừ đi 1 để lấy giá trị <= r
        left_idx = bisect.bisect_left(q_vals, l)
        right_idx = bisect.bisect_right(q_vals, r) - 1
        
        # Duyệt qua các query lọt vào trong phạm vi này
        for j in range(left_idx, right_idx + 1):
            orig_idx = sorted_queries[j][1]
            
            # Nếu query này CHƯA có đáp án (đang là -1), ta cập nhật luôn.
            # Nhờ intervals đã sắp xếp theo độ dài, đây CHẮC CHẮN là độ dài ngắn nhất.
            if ans[orig_idx] == -1:
                ans[orig_idx] = length
                
    return ans

# --- Chạy thử kiểm tra với ví dụ 1 trong ảnh ---
intervals_1 = [[1, 4], [2, 4], [3, 6], [4, 4]]
queries_1 = [2, 3, 4, 5]
print(minInterval(intervals_1, queries_1)) 
# Kết quả mong đợi: [3, 3, 1, 4]
```

### W7A9
#### Cho một danh sách các số nums, với mỗi phần tử nums[i], hãy đếm xem có bao nhiêu phần tử bên phải nhỏ hơn nums[i].
#### Trả về danh sách các số đếm.
#### Efficient solution: using heap or binary search tree

```python
import bisect
def h(nums):
    ans=[]
    seen=[]
    for num in reversed(nums):
        vitri=bisect.bisect_left(seen,num)
        ans.append(vitri)
        bisect.insort(seen,num)
    return ans[::-1]
nums=list(map(int,input().split()))
print(h(nums))
```

### W7A10
#### Cho một danh sách các số nguyên nums và một số nguyên m, hãy tìm tổng mảng con lớn nhất theo modulo m.
#### Tức là, với bất kỳ mảng con nums[i:j] nào, hãy tính sum(nums[i:j]) % m và trả về kết quả lớn nhất có thể.
#### Gợi ý:
#### Duy trì tổng tiền tố (prefix sum) mod m và sử dụng tìm kiếm nhị phân để tìm tổng tiền tố nhỏ nhất chỉ lớn hơn tổng hiện tại một chút (để tối đa hóa hiệu số bao quanh).


```python
import bisect

def max_subarray_mod(nums, m):
    max_mod = 0
    prefix_sum = 0
    seen = []  # Giỏ chứa các tổng tiền tố đã gặp, luôn sắp xếp tăng dần
    
    for num in nums:
        # 1. Tính tổng tiền tố hiện tại và lấy dư luôn cho m
        prefix_sum = (prefix_sum + num) % m
        
        # 2. Trường hợp cơ bản: Đoạn từ đầu mảng đến vị trí hiện tại
        max_mod = max(max_mod, prefix_sum)
        
        # 3. Dùng tìm kiếm nhị phân để tìm giá trị NHỎ NHẤT nhưng LỚN HƠN prefix_sum hiện tại
        # bisect_right giúp bỏ qua các số bằng prefix_sum, tìm số lớn hơn hẳn
        idx = bisect.bisect_right(seen, prefix_sum)
        
        # 4. Nếu tìm thấy một số như vậy trong giỏ 'seen'
        if idx < len(seen):
            # Tính hiệu số vòng quanh (wrap-around)
            candidate = prefix_sum - seen[idx] + m
            # Cập nhật kỷ lục nếu giá trị mới này lớn hơn
            max_mod = max(max_mod, candidate)
            
        # 5. Nhét tổng tiền tố hiện tại vào giỏ để làm vốn cho các bước sau
        bisect.insort(seen, prefix_sum)
        
    return max_mod
# --- PHẦN NHẬP LIỆU ĐỂ NỘP BÀI ---
if __name__ == '__main__':
    # Đọc dòng đầu tiên (chứa mảng). Dùng eval() để đọc mảng có dấu ngoặc vuông []
    nums = eval(input())
    
    # Đọc dòng thứ hai (chứa số m)
    m = int(input())
    
    # In ra đáp án cuối cùng
    print(max_subarray_mod(nums, m))
```

### W7A13
#### Dãy hài hòa là một dãy thỏa mãn điều kiện giá trị lớn nhất và giá trị nhỏ nhất của dãy khác nhau đúng bằng \( 1 \).
#### Viết hàm trả về độ dài của dãy hài hòa con dài nhất có thể có trong mảng giá trị nguyên \( arr \) có \( n \) phần tử
#### Ví dụ:
#### Input: 1 3 2 2 5 2 3 7
#### Output: 5 - Dãy hài hòa con dài nhất là [3,2,2,2,3]


```python
def h(arr):
    demso={}
    for x in arr:
        if x in demso:
            demso[x]+=1
        else:
            demso[x]=1
    maxlen=0
    for x in demso:
        if x+1 in demso:
            chieudai=demso[x]+demso[x+1]
            maxlen=max(maxlen,chieudai)
    return maxlen
arr=list(map(int,input().split()))
print(h(arr))
```

#### Nếu không tìm được số \( 7 \) nào in ra "Not found" 

```python
def h(n,arr):
    vitri=[]
    for i in range(n-1,-1,-1):
        if arr[i]==7:
            vitri.append(str(i))
    if len(vitri)==0:
        print('Not found')
    else:
        print(' '.join(vitri))
n=int(input())
arr=list(map(int,input().split()))
h(n,arr)
```
