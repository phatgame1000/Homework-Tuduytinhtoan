### W6A16
#### Nhâ ̣p vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣ trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

```python
n,m=map(int,input().split())
mt=[]
for i in range(n):
    h=list(map(int,input().split()))
    mt.append(h)
for i in range(n):
    for j in range(m):
        print(f'{mt[i][j]:>4}',end='')
    print()
```

### W6A17
#### Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và đường chéo phụ của mảng

```python
n=int(input())
mt=[]
for i in range(n):
    h=list(map(int,input().split()))
    mt.append(h)
for i in range(n):
    for j in range(n):
        if i==j or i+j==n-1:
            print(f'{mt[i][j]:>3}',end=' ')
        else:
            print('   ',end=' ')
    print()
```

### W6A19
#### Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất trong list

```python
def h(n:list)->int:
    mx=0
    for i in range(1,len(n)):
        if n[i]>n[mx]:
            mx=i
    return mx
n=list(map(int,input().split()))
print(h(n))
```

### W6A20
#### Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về -1


```python
def h(n:list,k:int)->int:
    vt=-1
    for i in range(len(n)):
        if n[i]==k:
            return i
    return vt
n=list(map(int,input().split()))
k=int(input())
print(h(n,k))
```

### W6A21
#### Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
n=input().split()
z={}
for i in range(len(n)):
    x=n[i]
    if x not in z:
        z[x]=[i]
    else:
        z[x].append(i)
for x in z:
    z[x]=tuple(z[x])
print(z)
```

### W6A24
#### Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9


```python
def h(n:list)->list:
    khackhong=[]
    bangkhong=[]
    for i in n:
        if i!=0:
            khackhong.append(i)
        else:
            bangkhong.append(0)
    return bangkhong+khackhong
n=list(map(int,input().split()))
print(*h(n))
```

### W6A26
#### Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 phần tử đầu và cuối

```python
n=list(map(float,input().split()))
tong=0
for i in range(1,len(n)-1):
    if n[i]>n[i-1] and n[i]>n[i+1]:
        tong+=n[i]
print(tong)
```

### W6A27
#### Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước). nếu không có in ra -1

```python
*n,x=list(map(int,input().split()))
z=[]
for i in range(len(n)):
    if n[i]<x:
        z.append(i)
if not z:
    print(-1)
else:
    print(*z)
```

### W6A28
#### Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm thực sự, hay vô danh (không tăng, không giảm)

```python
n=list(map(int,input().split()))
tang=True
giam=True
for i in range(1,len(n)):
    if n[i]<=n[i-1]:
        tang=False
    if n[i]>=n[i-1]:
        giam=False
if tang:
    print('Dãy tăng thực sự')
elif giam:
    print('Dãy giảm thực sự')
else:
    print('Vô danh')
```
