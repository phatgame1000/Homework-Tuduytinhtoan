### W7A16
#### Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu mỗi phần tử x trong mảng xuất hiện đúng x lần.
#### Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.


```python
def h(arr):
    demso={}
    for x in arr:
        if x in demso:
            demso[x]+=1
        else:
            demso[x]=1
    tongsoxoa=0
    for x,c in demso.items():
        if c>=x:
            tongsoxoa+=(c-x)
        else:
            tongsoxoa+=c
    if tongsoxoa==0:
        return 'YES'
    else:
        return f'NO {tongsoxoa}'
arr=list(map(int,input().split()))
print(h(arr))
```

```python
def h(arr):
    n=len(arr)
    if n==0:
        return
    a=None
    b=0
    for x in arr:
        if b==0:
            a=x
            b=1
        elif x==a:
            b+=1
        else:
            b-=1
    c=0
    for x in arr:
        if x==a:
            c+=1
    if c>n/2:
        print(a)
arr=list(map(int,input().split()))
h(arr)
```

### W7A25
#### Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số


```python
def h(num_str,k):
    stack=[]
    drop_count=len(num_str)-k
    for digit in num_str:
        while stack and drop_count>0 and stack[-1]<digit:
            stack.pop()
            drop_count-=1
        stack.append(digit)
    return ''.join(stack[:k])
data=input().split()
num_str=data[0]
k=int(data[1])
print(h(num_str,k))
```

### W7A26
#### Tách các số có trong một đoạn văn bản, tính tổng và tính trung bình các số đó.

```python
def h(text):
    chuoisach=''
    for char in text:
        if char.isdigit():
            chuoisach+=char
        else:
            chuoisach+=' '
    cumso=chuoisach.split()
    danhsachso=[]
    for so in cumso:
        danhsachso.append(int(so))
    if not danhsachso:
        return 0
    tong=sum(danhsachso)
    trungbinh=tong/len(danhsachso)
    return trungbinh
text=input()
print(h(text))
```

### W7A27
#### Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ hay không


```python
def h(s):
    stack=[]
    batcap={')':'(','}':'{',']':'['}
    for char in s:
        if char in batcap:
            top_element=stack.pop() if stack else '#'
            if batcap[char]!=top_element:
                return False
        else:
            stack.append(char)
    return not stack
s=input()
print(h(s))
```

### W7A30
#### Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập vào từ bàn phím 

```python
def tim_day_tang_khong_lien_tuc(arr):
    n = len(arr)
    if n == 0:
        return []
        
    # dp[i]: Lưu độ dài dãy tăng dài nhất kết thúc tại vị trí i. Khởi tạo mặc định là 1 (chính nó)
    dp = [1] * n
    
    # trace[i]: Lưu lại "người đứng trước" của phần tử i để lát nữa truy vết. -1 nghĩa là không có ai.
    trace = [-1] * n 
    
    # Bước 1: Quy hoạch động - Duyệt qua từng cặp số
    for i in range(1, n):
        for j in range(0, i):
            # Nếu số đứng sau > số đứng trước VÀ ghép vào giúp dãy dài ra thêm
            if arr[i] > arr[j] and dp[i] < dp[j] + 1:
                dp[i] = dp[j] + 1 # Cập nhật kỷ lục mới cho vị trí i
                trace[i] = j      # Ghi nhớ lại: "Phần tử i vừa ghép nối đằng sau phần tử j"
                
    # Bước 2: Tìm vị trí của điểm kết thúc dãy dài nhất
    max_len = 0
    end_index = 0
    for i in range(n):
        if dp[i] > max_len:
            max_len = dp[i]
            end_index = i
            
    # Bước 3: Lần theo mảng trace để lấy lại dãy số thực tế
    day_con = []
    while end_index != -1:
        day_con.append(arr[end_index])
        end_index = trace[end_index] # Lùi về người đứng trước
        
    # Vì ta đang truy ngược từ cuối lên đầu, nên mảng kết quả đang bị ngược. Cần đảo lại.
    return day_con[::-1]

# --- Phần chạy thử nghiệm ---
try:
    # Nhập ví dụ: 10 22 9 33 21 50 41 60 80
    danh_sach = list(map(int, input("Nhập danh sách: ").split()))
    
    ket_qua = tim_day_tang_khong_lien_tuc(danh_sach)
    print("Dãy con tăng dần dài nhất:", ket_qua)
except ValueError:
    print("Vui lòng chỉ nhập các chữ số hợp lệ.")
```
