### Bài 10:
#### Nhập vào số nguyên dương a, kiểm tra xem a có phải là số hoàn hảo hay không. Số hoàn hảo là số có tổng các ước (ước nhỏ hơn nó) bằng chính nó, VD: số 6, 28… 

```python
a=int(input())
if a<=1:
    print(False)
i=2
b=1
while i*i<=a-1:
    if a%i==0:
        b+=i
        j=a//i
        if j!=i:
            b+=j
    i+=1
if b==a:
    print(True)
else:
    print(False)
```

### Bài 11:

### Bài 12:
#### Nhập số nguyên dương n, Cho S(k) = 1 + 2 + 3 + … + k. Tìm k sao cho S(k) lớn nhất nhưng nhỏ hơn n 


```python
n=int(input())
k=0
S=0
while S+(k+1)<n:
    k+=1
    S+=k
print(k)
```

### Bài 14:
#### Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. KHÔNG ĐƯỢC dùng mảng 

```python
l=None
n=None
try:
    b=input().split()
    for c in b:
        a=int(c)
        if a==-1:
            break
        if l is None and n is None:
            l=a
            n=a
        else:
            if a>l:
                l=a
            if a<n:
                n=a
except ValueError:
    pass
if l is not None:
    print(f'{l} {n}')
    
```

### Bài 19:
#### Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt quá A 


```python
a=int(input())
b=1
c=1
while c<=a:
    b,c=c,b+c
print(b)
```

#### Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố.
