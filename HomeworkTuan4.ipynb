```python
n=int(input())
s=0
k=0
while s+(k+1)<n:
    k+=1
    s+=k
print(k)
```

### W4A25
#### Nhập vào một dãy số nguyên, ngưng nhập khi người dùng nhập -1.
#### Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập 


```python
l=None
n=None
try:
    b=input().split()
    for c in b:
        a=int(c)
        if a==-1:
            break
        if l is None and n is None:
            l=a
            n=a
        else:
            if a>l:
                l=a
            if a<n:
                n=a
except ValueError:
    pass
if l is not None:
    print(f'{l} {n}')
```

### W4A26
#### [Fibonaci1]
#### Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ bằng tổng hai số trước đó Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt quá A 

```python
a=int(input())
b=1
c=1
while c<=a:
    b,c=c,b+c
print(b)    
```

### W4A28
#### Nhập vào một chuỗi, hãy in từ đầu tiên trong chuỗi
#### Dùng cách: for trên index; KHÔNG dùng hàm split() 

```python
a=input()
b=''
for i in range(len(a)):
    if a[i]!=' ':
        b+=a[i]
    else:
        if b!='':
            break
print(b)
```

### W4A29
#### Cho một chuỗi chỉ gồm các ký tự chữ cái thường và chữ cái hoa, hãy đổi ký tự hoa thành thường và thường thành hoa. 

```python
n=input()
a=''
for i in n:
    if i.islower():
        a+=i.upper()
    elif i.isupper():
        a+=i.lower()
    else:
        a+=i
print(a)
```

### W4A31
#### Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu ký tự in hoa, bao nhiêu ký tự in thường, bao nhiêu ký tự số


```python
n=input()
a=0
b=0
c=0
for i in n:
    if 'a'<=i<='z':
        b+=1
    elif 'A'<=i<='Z':
        a+=1
    elif '0'<=i<='9':
        c+=1
print(a,b,c)
```
