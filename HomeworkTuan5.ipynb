### W5A16
#### Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x

```python
def a(x:int)->bool:
    if x<0:
        return False
    y=x
    n=0
    while y>0:
        n+=1
        y//=10
    y=x
    m=0
    while y>0:
        z=y%10
        m+=z**n
        y//=10
    return m==x
x=int(input())
print(a(x))
```

#### Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.

```python
def h(x:str)->str:
    try:
        int(x)
        return 'int'
    except ValueError:
        pass
    try:
        float(x)
        return 'float'
    except ValueError:
        pass
    return 'neither'
x=input()
print(h(x))
```

### W5A25
#### Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3), (0,3)], p = (2,1)


```python
def point_in_quadrilateral(quad:list,p:tuple)->bool:
    def h(a,b,c):
        return (b[0]-a[0])*(c[1]-a[1])-(c[0]-a[0])*(b[1]-a[1])
    A,B,C,D=quad
    h1=h(A,B,p)
    h2=h(B,C,p)
    h3=h(C,D,p)
    h4=h(D,A,p)
    fullduong=h1>=0 and h2>=0 and h3>=0 and h4>=0
    fullam=h1<=0 and h2<=0 and h3<=0 and h4<=0
    return fullduong or fullam
n=list(map(float,input().split()))
quad=[(n[i],n[i+1])for i in range(0,8,2)]
p=(n[8],n[9])
print(point_in_quadrilateral(quad,p))
```

### W5A27
#### Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list)

```python
def h(n:list)->int:
    y=0
    for m in n:
        x=abs(m)
        if x==0:
            continue
        while x>0:
            z=x%10
            if z!=0:
                y+=1
            x//=10
    return y
n=list(map(int,input().split()))
print(h(n))
```

### W5A28
#### Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).

```python
import math
def h(x1,y1,x2,y2,x3,y3)->tuple:
    def k(x1,y1,x2,y2,x3,y3):
        vx,vy=x2-x1,y2-y1
        ux,uy=x3-x1,y3-y1
        a=math.hypot(vx,vy)
        b=math.hypot(ux,uy)
        c=vx*ux+vy*uy
        if a==0 or b==0:
            return 0.0
        d=c/(a*b)
        cos=max(-1.0,min(1.0,d))
        return math.degrees(math.acos(cos))
    A=k(x1,y1,x2,y2,x3,y3)
    B=k(x2,y2,x1,y1,x3,y3)
    C=k(x3,y3,x1,y1,x2,y2)
    return A,B,C
n=list(map(float,input().split()))
A,B,C=h(*n)
print(f'A={A:.2f}')
print(f'B={B:.2f}')
print(f'C={C:.2f}')
```

### W5A30
#### Viết hàm tính hệ số nhị thức C(n, k).


```python
import math
def h(n:int,k:int)->int:
    if k<0 or k>n:
        return 0
    a=math.factorial(n)
    b=math.factorial(k)*math.factorial(n-k)
    return a//b
n,k=map(int,input().split())
print(h(n,k))
```

### W5A33
#### Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa, giá km, phí chờ, phụ phí ban đêm và giảm giá.

Loại xe (car_type): 4 chỗ hoặc 7 chỗ.

Giá Km: Dưới 1km (giá mở cửa), từ km 2 đến 30 (giá thường), từ km 31 trở đi (giá đường dài).
    
    4 chỗ : 20000, 15000, 12000
    
    7 chỗ : 25000, 17000, 14000

Phí chờ (wait_minutes): Miễn phí 5 phút đầu, sau đó tính tiền 1000/1 phút

Phụ phí đêm (start_hour): Từ 22h đêm đến 5h sáng bị cộng thêm 20%.

Giảm giá: Đi xa trên 50km được giảm 10% tổng hóa đơn.

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    t=0
    if car_type==4:
        a=20000
        b=15000
        c=12000
    elif car_type==7:
        a=25000
        b=17000
        c=14000
    else:
        return 0
    if km<=1:
        t=a
    elif km<=30:
        t=a+(km-1)*b
    else:
        t=a+29*b+(km-30)*c
    if wait_minutes >5:
        d=(wait_minutes-5)*1000
        t+=d
    if start_hour>=22 or start_hour<=5:
        t=t*1.2
    if km>50:
        t*=0.9
    return t
car_type, start_hour, km, wait_minutes=map(float,input().split())
print(smart_taxi_fare(car_type, start_hour, km, wait_minutes))
```

### W5A34
#### Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau Nếu n ≤ 0 trả về 0.

```python
def sum_1_to_n(n:int)->int:
    if n<=0:
        return 0
    t=0
    for x in range(1,n+1):
        y=str(x)
        if len(y)==len(set(y)):
            t+=x
    return t
n=int(input())
print(sum_1_to_n(n))
```
